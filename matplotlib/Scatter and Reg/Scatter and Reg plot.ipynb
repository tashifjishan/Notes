```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("exam.csv")
```

```python
df
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Create figure and axes
fig, ax = plt.subplots(figsize=(10, 6), facecolor="#121212")

# Regression calculation
m, c = np.polyfit(df["Study_Hours"], df["Exam_Score"], 1)

# Generate smooth regression line
x_line = np.linspace(
    df["Study_Hours"].min(),
    df["Study_Hours"].max(),
    100
)

y_line = m * x_line + c

# Plot regression line
ax.plot(
    x_line,
    y_line,
    color="crimson",
    linewidth=3,
    label=f"Regression Line: y={m:.2f}x+{c:.2f}"
)

# Scatter plot
scatter = ax.scatter(
    df["Study_Hours"],
    df["Exam_Score"],
    s=df["Exam_Score"]*2,
    c=df["Study_Hours"],
    cmap="coolwarm",
    alpha=0.75,
    marker="^",
    edgecolors="white",
    linewidths=0.8,
    label="Students"
)

# Background
ax.set_facecolor("#121212")

# Labels and title
ax.set_title(
    "Study Hours vs Exam Score",
    fontsize=18,
    fontweight="bold",
    color="white",
    pad=15
)

ax.set_xlabel(
    "Study Hours",
    fontsize=12,
    color="white"
)

ax.set_ylabel(
    "Exam Score",
    fontsize=12,
    color="white"
)

# Tick styling
ax.tick_params(
    axis="both",
    colors="white",
    labelsize=11
)

# Spine styling
for spine in ax.spines.values():
    spine.set_color("white")
    spine.set_linewidth(1)

# Grid
ax.grid(
    color="gray",
    linestyle="--",
    linewidth=0.5,
    alpha=0.3
)

# Colorbar
cbar = plt.colorbar(scatter, ax=ax)
cbar.set_label(
    "Study Hours",
    color="white",
    fontsize=11
)
cbar.ax.tick_params(colors="white")

# Legend
legend = ax.legend(
    facecolor="#121212",
    edgecolor="white"
)

for text in legend.get_texts():
    text.set_color("white")

plt.tight_layout()

plt.show()
```

| Parameter    | Purpose                     | Example               |
| ------------ | --------------------------- | --------------------- |
| `x`          | X-axis values               | `x=df["Study_Hours"]` |
| `y`          | Y-axis values               | `y=df["Exam_Score"]`  |
| `s`          | Marker size                 | `s=80`                |
| `c`          | Marker color                | `c="red"`             |
| `marker`     | Marker shape                | `marker="^"`          |
| `alpha`      | Transparency                | `alpha=0.7`           |
| `edgecolors` | Border color                | `edgecolors="black"`  |
| `linewidths` | Border width                | `linewidths=2`        |
| `label`      | Legend label                | `label="Students"`    |
| `cmap`       | Colormap for numeric colors | `cmap="viridis"`      |
