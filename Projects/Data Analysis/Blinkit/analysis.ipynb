## 1. Sales Performance

```python
# Total Sales across all products: 

print(f'Total Sales: {data["Sales"].sum():,.2f}')
```

```python
# Which product generates the highest Sale:
data.groupby("Item Type")["Sales"].sum()
```

```python
# How are sales distributed across products?
fig, ax = plt.subplots(figsize=(5, 5))
data_g_itemtype = data.groupby("Item Type")[["Sales"]].sum().sort_values("Sales").reset_index()
ax.barh(data_g_itemtype["Item Type"], data_g_itemtype["Sales"], color="orange")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
# ax.spines["bottom"].set_visible(False)
# ax.spines["left"].set_visible(False)
```
